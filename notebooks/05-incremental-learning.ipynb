# **Определение и сравнение с классической моделью**
В **классическом машинном обучении** модель обучается один раз на исторических данных и затем используется в "статическом" режиме. В реальном мире данные часто поступают потоком (стримом), и распределение данных может меняться со временем.

**Инкрементальное обучение** (или online learning) — это парадигма, при которой модель обучается последовательно, получая данные по одному или небольшими порциями (батчами), и обновляет свои параметры без необходимости переобучаться на всех данных с нуля.

| Batch                           | Инкрементальное                     |
| ------------------------------- | ----------------------------------- |
| Обучение на всём датасете сразу | Обучение по мере поступления данных |
| Модель статична после обучения  | Модель постоянно обновляется        |
| Требует пересчёта с нуля        | Обновляется частично                |
| Хорошо при стабильных данных    | Нужно при изменяющейся среде        |


### Основные вызовы:
1.  **Catastrophic forgetting (катастрофическое забывание)** - модель может "забывать" старые паттерны при обучении на новых данных.
2.  **Concept drift (дрейф концепта)** - в реальной жизни происходит изменение распределения данных или целевой переменной со временем.
3.  **Resource constraints** - ограниченная память и вычислительные ресурсы.

### Примеры применения инкрементального обучения:
1. Интернет-магазин обновляет рекомендации по мере появления новых покупок;
2. Банк адаптирует модель скоринга по мере изменения поведения клиентов;
3. Антифрод-система учится на новых типах мошенничества.


## **Основные сценарии инкрементального обучения**
**Online Learning (обучение на потоке данных)** – модель обновляется после каждого нового примера или небольшого батча.

**Continual Learning (непрерывное обучение)** – модель обучается на новых задачах, не забывая старые.

**Transfer + Incremental**– комбинация: предобученная модель + дообучение на новых данных

## **Основные подходы инкрементального обучения**
### Fine-tuning
1. **Fine-tuning** – дообучение модели на новых данных с маленькой скоростью обучения.

*Плюсы:*

* Быстро

*Минусы:*
* Возможное забывание старых паттернов

### Replay
2. **Replay (повтор старых данных)**– при дообучении используется часть старых данных, мешаются старые и новые примеры – это снижает catastrophic forgetting.

3. **Регуляризационные методы** – добавляют штраф за изменение важных параметров модели. (модель может учиться новому, но не сильно меняет важные веса)

### Модульные архитектуры

# **Практическая реализация**
Давайте сгенерируем данные и дрейфом концепта

**Зачем?**

Чтобы показать разницу между подходами обучения:

* Классическая модель: выдает ошибочный результат, тк не подстроилась под изменение

* Инкрементальная модель: подстроилась и работает хорошо

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, SGDRegressor
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')


# 1. ГЕНЕРИРУЕМ ДАННЫЕ С ДРЕЙФОМ КОНЦЕПТА
np.random.seed(42)

# Создаем 500 точек данных, которые постепенно меняют свой паттерн
n_points = 500
X = np.linspace(0, 20, n_points).reshape(-1, 1)

# Первая половина данных - одна зависимость
y1 = 2 * X[:250] + 5 + np.random.normal(0, 2, (250, 1))

# Вторая половина - зависимость меняется (дрейф концепта!)
y2 = 0.5 * X[250:] + 15 + np.random.normal(0, 2, (250, 1))

# Объединяем
y = np.vstack([y1, y2])


plt.figure(figsize=(14, 5))
plt.scatter(X[:250], y[:250], alpha=0.6, label='Первая эпоха (сильная зависимость)', color='blue')
plt.scatter(X[250:], y[250:], alpha=0.6, label='Вторая эпоха (слабая зависимость)', color='red')
plt.axvline(x=10, color='black', linestyle='--', label='Момент изменения (дрейф)')
plt.xlabel('Признак X')
plt.ylabel('Целевая переменная Y')
plt.title('Рисунок 1: Данные с дрейфом концепта (изменение зависимости)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("До пунктира зависимость была сильной (крутой наклон), после - слабой (почти горизонтально)")
print("- это изменение данных в динамике, которое часто происходит в реальном времени")
```

# Классическая модель
1. Собираем все данные
2. Обучаем модель один раз
3. Используем её вечно

```python
# 2. КЛАССИЧЕСКИЙ ПОДХОД (BATCH LEARNING)
# Обучаемся один раз на ВСЕХ данных и замираем

# Масштабируем признаки
scaler_classic = StandardScaler()
X_scaled_classic = scaler_classic.fit_transform(X)

# Обучаем модель на ВСЕХ данных сразу
classic_model = LinearRegression()
classic_model.fit(X_scaled_classic, y)

# Предсказания для всех точек
y_pred_classic = classic_model.predict(X_scaled_classic)

# Считаем ошибку на разных участках
error_classic_first = np.mean((y[:250] - y_pred_classic[:250])**2)
error_classic_second = np.mean((y[250:] - y_pred_classic[250:])**2)

# Визуализация классического подхода
plt.figure(figsize=(14, 6))

# Данные
plt.scatter(X[:250], y[:250], alpha=0.3, color='blue', label='Данные (1 эпоха)')
plt.scatter(X[250:], y[250:], alpha=0.3, color='red', label='Данные (2 эпоха)')

# Предсказания классической модели
plt.plot(X, y_pred_classic, color='green', linewidth=3, label='Классическая модель (обучена 1 раз на всех данных)')
plt.axvline(x=10, color='black', linestyle='--', alpha=0.5)

plt.xlabel('Признак X')
plt.ylabel('Целевая переменная Y')
plt.title('Рисунок 2: Классический подход - модель застыла во времени')
plt.legend()
plt.grid(True, alpha=0.3)

# Добавляем текст с ошибками
plt.text(2, 30, f'Ошибка на 1 эпохе: {error_classic_first:.2f}',
         bbox=dict(boxstyle="round,pad=0.3", facecolor="blue", alpha=0.2))
plt.text(12, 30, f'Ошибка на 2 эпохе: {error_classic_second:.2f}',
         bbox=dict(boxstyle="round,pad=0.3", facecolor="red", alpha=0.2))

plt.show()

print("⚠️ Проблема: классическая модель пытается найти один усредненный паттерн для ВСЕХ данных")
print(f"   Из-за этого она ошибается и в начале (ошибка {error_classic_first:.2f}), и в конце (ошибка {error_classic_second:.2f})")
```

# Инкрементальный подход

Модель учится постоянно

1. Получил новые данные
2. Предсказал
3. Увидел правильный ответ
4. Чуть-чуть подстроился
5. Повторил


```python
# 3. ИНКРЕМЕНТАЛЬНЫЙ ПОДХОД (ONLINE LEARNING)
# Обучаемся постепенно, адаптируясь к новым данным

scaler_inc = StandardScaler()
# Важно: для инкрементального обучения мы масштабируем данные ПО МЕРЕ поступления
# Но для простоты примера используем то же масштабирование

# Создаем инкрементальную модель (SGD = Stochastic Gradient Descent)
inc_model = SGDRegressor(max_iter=1, tol=None, learning_rate='adaptive', eta0=0.01)

# Будем обучаться последовательно и сохранять предсказания
y_pred_inc = np.zeros_like(y)
errors_over_time = []

# Обучаемся ПО ОДНОЙ точке за раз (инкрементально!)
for i in range(len(X)):
    # Масштабируем текущую точку
    X_current = scaler_classic.transform(X[i:i+1])  # используем тот же scaler для простоты

    if i == 0:
        # Первый раз - нужно указать классы
        inc_model.partial_fit(X_current, y[i].ravel())
        y_pred_inc[i] = y[i]  # для первой точки нет предсказания
    else:
        # Предсказываем ДО обучения на этой точке
        y_pred_inc[i] = inc_model.predict(X_current)
        # Обучаемся на этой точке ПОСЛЕ предсказания
        inc_model.partial_fit(X_current, y[i].ravel())

    # Считаем среднюю ошибку за последние 50 шагов
    if i >= 50:
        errors_over_time.append(np.mean((y[i-50:i].ravel() - y_pred_inc[i-50:i].ravel())**2))
    else:
        errors_over_time.append(0)

# Визуализация инкрементального подхода
plt.figure(figsize=(14, 10))

# График 1: Данные и предсказания
plt.subplot(2, 1, 1)
plt.scatter(X[:250], y[:250], alpha=0.3, color='blue', label='Данные (1 эпоха)', s=20)
plt.scatter(X[250:], y[250:], alpha=0.3, color='red', label='Данные (2 эпоха)', s=20)

# Предсказания инкрементальной модели (цвет меняется со временем)
scatter = plt.scatter(X[1:], y_pred_inc[1:], c=range(1, n_points),
                      cmap='viridis', alpha=0.8, s=30, label='Предсказания модели', vmin=0, vmax=n_points)
plt.colorbar(scatter, label='Время (номер точки)')
plt.axvline(x=10, color='black', linestyle='--', alpha=0.5)

plt.xlabel('Признак X')
plt.ylabel('Целевая переменная Y')
plt.title('Рисунок 3: Инкрементальный подход - модель эволюционирует')
plt.legend(loc='upper left')
plt.grid(True, alpha=0.3)

# График 2: Ошибка модели со временем
plt.subplot(2, 1, 2)
plt.plot(range(len(errors_over_time)), errors_over_time, linewidth=2, color='purple')
plt.axvline(x=250, color='red', linestyle='--', alpha=0.7, label='Момент дрейфа')
plt.xlabel('Время (номер точки)')
plt.ylabel('Средняя ошибка (MSE) за последние 50 точек')
plt.title('Рисунок 4: Как меняется ошибка инкрементальной модели во времени')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("   - Сначала модель училась на синей зависимости (ошибка падала)")
print("   - В момент дрейфа (красная линия) ошибка резко выросла - модель растерялась")
print("   - Но потом она быстро адаптировалась к новой реальности (ошибка снова упала)")
```

Теперь давайте посмотрим, как происходит обработка данных в процессе

```python
# 4. СРАВНЕНИЕ В ДИНАМИКЕ

# Разобьем данные на 5 временных периодов и посмотрим на ошибки каждой модели
n_periods = 5
period_size = n_points // n_periods

classic_errors_periods = []
inc_errors_periods = []

for p in range(n_periods):
    start = p * period_size
    end = (p + 1) * period_size if p < n_periods - 1 else n_points

    # Ошибка классической модели
    classic_errors_periods.append(np.mean((y[start:end] - y_pred_classic[start:end])**2))

    # Ошибка инкрементальной модели
    inc_errors_periods.append(np.mean((y[start:end].ravel() - y_pred_inc[start:end].ravel())**2))

# Визуализация сравнения
plt.figure(figsize=(14, 6))

periods = ['Период 1', 'Период 2', 'Период 3', 'Период 4', 'Период 5']
x_pos = np.arange(len(periods))
width = 0.35

plt.bar(x_pos - width/2, classic_errors_periods, width, label='Классическая модель', color='steelblue', alpha=0.8)
plt.bar(x_pos + width/2, inc_errors_periods, width, label='Инкрементальная модель', color='coral', alpha=0.8)

# Отмечаем где произошел дрейф
plt.axvline(x=1.5, color='red', linestyle='--', linewidth=2, label='Момент дрейфа концепта')

plt.xlabel('Временной период')
plt.ylabel('Средняя ошибка (MSE)')
plt.title('Рисунок 5: Сравнение классического и инкрементального подхода')
plt.xticks(x_pos, periods)
plt.legend()
plt.grid(True, alpha=0.3)

# Добавляем подписи значений
for i, (classic_err, inc_err) in enumerate(zip(classic_errors_periods, inc_errors_periods)):
    plt.text(i - width/2, classic_err + 0.5, f'{classic_err:.1f}', ha='center', va='bottom', fontsize=9)
    plt.text(i + width/2, inc_err + 0.5, f'{inc_err:.1f}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()

print("\n📊 ИТОГОВОЕ СРАВНЕНИЕ:")
print("="*50)
print("Классическая модель:")
print(f"  - Ошибка до дрейфа: {classic_errors_periods[0]:.2f}")
print(f"  - Ошибка после дрейфа: {classic_errors_periods[3]:.2f}")
print(f"  - Изменение: {classic_errors_periods[3] - classic_errors_periods[0]:.2f}")
print()
print("Инкрементальная модель:")
print(f"  - Ошибка до дрейфа: {inc_errors_periods[0]:.2f}")
print(f"  - Ошибка сразу после дрейфа: {inc_errors_periods[3]:.2f}")
print(f"  - Ошибка спустя время (адаптация): {inc_errors_periods[4]:.2f}")
print(f"  - Улучшение после адаптации: {inc_errors_periods[3] - inc_errors_periods[4]:.2f}")
print("="*50)
```

```python
# 5. БОНУС: АНИМАЦИЯ (почти) - как инкрементальная модель "переучивается"

# Покажем, как меняются веса (коэффициенты) модели со временем
weights_over_time = []

# Переобучим с нуля, но теперь будем сохранять веса
inc_model_weight = SGDRegressor(max_iter=1, tol=None, learning_rate='adaptive', eta0=0.01)

for i in range(len(X)):
    X_current = scaler_classic.transform(X[i:i+1])

    if i == 0:
        inc_model_weight.partial_fit(X_current, y[i].ravel())
    else:
        inc_model_weight.partial_fit(X_current, y[i].ravel())

    # Сохраняем вес (коэффициент) модели
    weights_over_time.append(inc_model_weight.coef_[0])

plt.figure(figsize=(14, 5))
plt.plot(range(len(weights_over_time)), weights_over_time, linewidth=2, color='darkorange')
plt.axvline(x=250, color='red', linestyle='--', linewidth=2, label='Момент дрейфа')
plt.axhline(y=2, color='blue', linestyle=':', alpha=0.7, label='Истинный вес в 1 эпохе')
plt.axhline(y=0.5, color='red', linestyle=':', alpha=0.7, label='Истинный вес во 2 эпохе')
plt.xlabel('Время (номер точки)')
plt.ylabel('Вес модели (коэффициент)')
plt.title('Рисунок 6: Как инкрементальная модель "переучивается" - изменение веса')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("Вес модели постепенно перестраивается с 2.0 на 0.5!")
```

## 6. **ВЫВОДЫ**

###Классический подход (Batch Learning)

**Плюсы:**
- Простой и понятный — обучил один раз и забыл
- Точно отражает прошлые данные — идеально помнит историю
- Не дергается от случайных выбросов — стабилен

**Минусы:**
- Не адаптируется к изменениям — застыл во времени
- Если мир поменялся — модель устарела и ошибается
- Нужно хранить все данные для переобучения — требует памяти

---

###Инкрементальный подход (Online Learning)

**Плюсы:**
- Адаптируется к изменениям — как мы видели в примере, вес модели перестроился с 2.0 на 0.5
- Не нужно хранить все данные — достаточно текущего состояния
- Работает в реальном времени — можно использовать для потоковых данных

**Минусы:**
- Может "забывать" старое — пластичность иногда побеждает стабильность
- Чувствителен к выбросам — одно резкое изменение может сильно дернуть модель
- Сложнее настраивать — нужно подбирать скорость обучения, размер окна и т.д.

---

### **Когда что использовать?**

| Ситуация | Какой подход |
|----------|--------------|
| Данные стабильны, можно переобучать раз в месяц | **Классический** |
| Данные постоянно меняются, нужна быстрая реакция | **Инкрементальный** |
| Мало данных, важна максимальная точность на истории | **Классический** |
| Данные приходят потоком (транзакции, логи, сенсоры) | **Инкрементальный** |
| Критична интерпретируемость, простота | **Классический** |
| Важна адаптация к новым условиям | **Инкрементальный** |

---

### **Что показал наш эксперимент?**

В нашем примере мы создали данные, которые меняются со временем:

1. **Сначала** работало правило: Y = 2×X + 5 (сильная зависимость от X)
2. **Потом** правило изменилось: Y = 0.5×X + 15 (слабая зависимость)

**Классическая модель** попыталась найти одну линию для всех данных и в результате:
- Недолетала в первой эпохе (линия слишком пологая)
- Перелетала во второй эпохе (линия слишком крутая)
- Так и застыла с этим усредненным представлением

**Инкрементальная модель**:
- Сначала выучила первую зависимость (вес около 2.0)
- В момент дрейфа растерялась (ошибка выросла)
- Но потом плавно перестроилась под новую реальность (вес сполз к 0.5)
- В итоге адаптировалась и снова стала хорошо работать

---

### **Главный вывод**

> В стабильном мире побеждает классический подход — он точнее помнит прошлое.
>
> В меняющемся мире побеждает инкрементальный — он успевает подстроиться под настоящее.

А чтобы понимать, насколько хорошо модель справляется с изменениями, нужны специальные метрики:
- **Prequential accuracy** — качество на свежих данных
- **Время реакции на дрейф** — скорость адаптации
- **Стабильность vs Пластичность** — баланс памяти и обучаемости
- **Память и скорость** — можно ли запустить на реальном устройстве

---

*В идеале — совмещать оба подхода: периодически переобучать инкрементальную модель на накопленных данных, чтобы освежать память о далеком прошлом.*

Метрики!


Prequential Accuracy

```python
# Считаем prequential accuracy с окном
window_size = 50
prequential_errors = []

for i in range(window_size, len(X)):
    # Берем последние window_size предсказаний
    window_pred = y_pred_inc[i-window_size:i].ravel()
    window_true = y[i-window_size:i].ravel()

    # Считаем ошибку
    mse = np.mean((window_true - window_pred)**2)
    prequential_errors.append(mse)

# Визуализируем
plt.figure(figsize=(14, 5))
plt.plot(range(window_size, len(X)), prequential_errors, linewidth=2, color='purple')
plt.axvline(x=250, color='red', linestyle='--', linewidth=2, label='Момент дрейфа')
plt.xlabel('Время (номер точки)')
plt.ylabel('MSE (средняя ошибка) за последние 50 точек')
plt.title('Рисунок 6: Prequential Accuracy - как модель работает СЕЙЧАС')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("""
📈 Что показывает этот график:
• Это НЕ средняя ошибка за всё время
• Это ошибка "прямо сейчас" (за последние 50 примеров)
• Видно, как модель:
  - Сначала учится (ошибка падает)
  - В момент дрейфа (красная линия) - шок! ошибка растет
  - Потом снова учится (ошибка падает)
""")
```

Время реакции на дрейф

```python
# Находим момент дрейфа и момент восстановления
drift_point = 250
recovery_point = None

# Ищем, когда ошибка вернулась к "нормальному" уровню
# Нормальный уровень = средняя ошибка ДО дрейфа + небольшой запас
normal_error_before = np.mean(prequential_errors[:150])  # первые 150 точек после окна
threshold = normal_error_before * 1.2  # +20% к норме

for i, err in enumerate(prequential_errors):
    if i + window_size > drift_point and err < threshold:
        recovery_point = i + window_size
        break

if recovery_point:
    reaction_time = recovery_point - drift_point
    print(f"⏱️  Время реакции на дрейф: {reaction_time} примеров")
    print(f"   • Модель заметила изменения и адаптировалась за {reaction_time} шагов")
    print(f"   • Это хороший показатель!")
else:
    print("❓ Модель так и не восстановилась (плохо!)")
```
